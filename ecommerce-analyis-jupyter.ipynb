```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Sample - Superstore.csv",encoding='latin-1')
```

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])
data['Ship Date'] = pd.to_datetime(data['Ship Date'])
```

# Creating new coloumns named Order Month, Order Year and Order day of the Month

```python
data['Order Month'] = data['Order Date'].dt.month
data['Order Year'] = data['Order Date'].dt.year
data['Order day of the month'] = data['Order Date'].dt.dayofweek
```

```python
data
```

```python
data.duplicated().sum()
```

```python
sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index()
max_sale = sales_by_month.max()
```

# Monthly Sale Analysis

```python
plt.figure(figsize=(10,6))
fig1=sns.lineplot(sales_by_month,x='Order Month', y='Sales')
plt.title("Monthly Sale Analysis",fontsize=12,fontweight='bold')
            
```

Analysis by Product Category

```python
sales_by_category=data.groupby('Category')['Sales'].sum().reset_index()
best_category=sales_by_category.max()
worst_category=sales_by_category.min()
```

# visual representation of category analysis

```python
fig2=plt.pie(sales_by_category['Sales'],
             labels=sales_by_category['Category'],
             startangle=90,
             wedgeprops={'width':0.7,'edgecolor':'white'},
             autopct='%1.1f%%')
plt.title("Sales Analysis through Category",fontsize=14,fontweight='bold')
fig2
```

# sales analysis by sub category

```python
sales_by_subcat = data.groupby('Sub-Category')['Sales'].sum().reset_index()
best_sub = sales_by_subcat.loc[sales_by_subcat['Sales'].idxmax()]
worst_sub = sales_by_subcat.loc[sales_by_subcat['Sales'].idxmin()]
```

```python
sales_by_subcat
```

```python
best_sub
```

```python
worst_sub
```

# visual representation of sub category analysis

```python
plt.figure(figsize=(8,10))

figure3 = sns.barplot(data=sales_by_subcat,x='Sales',y='Sub-Category')
plt.title("Sales Analysis through Sub-Category",fontsize=14,fontweight='bold')
for bars in figure3.containers:
    figure3.bar_label(bars)
figure3
```

# Sales to profit ratio

```python
sales_pro_sale = data.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
sales_pro_sale['sales_profit_ratio']=sales_pro_sale['Sales']/sales_pro_sale['Profit']
print(sales_pro_sale[['Segment','sales_profit_ratio']])
```
